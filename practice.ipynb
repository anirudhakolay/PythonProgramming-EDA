```python
#OOPS Abstraction: Hiding the implementation details of a class and only showing the essential features to the user.

class Car:
    def __inti__ (self):
        self.accelerate = False #not applied previously so false
        self.brake = False
        self.clutch = False 
    
    def start(self):
        self.accelerate = True
        self.brake = True
        self.clutch = True
        print("vroom vroom...")

car1 = Car()
car1.start() #only this method is exposed to the user
```

```python
#concrete class- Abstraction

from abc import ABC, abstractmethod
#abstract class
class vehicle(ABC): #inherites ABC
    @abstractmethod

    def start(self):
        pass
    def price(self):
        pass

#concrete class
class car(vehicle): #parent class
    def price(self):
        return "car price 2Lakh"
    def start(self):
        return "car started"

#object of concrete class
c1 = car()
print(c1.price())
print(c1.start())
```

```python
#inheritance
#Single inheritance
class Animal:
    def __init__(self,name, food):
        self.name = name
        self.food = food


class Dog(Animal):
    def __init__(self, name, food, breed, color, height): #name override
        super().__init__(name, food) #old from parents
        self.breed = breed
        self.color = color
        self.height = height

    def details(self):
        print(self.name)
        print(self.breed)
        print(self.color)
        print(self.food)
        print(self.height)

d1 = Dog("tommy", "dog food", "labrodor", "black", 5)
print(d1.details())

print(d1.name)
```

```python
#Multi-level Inheritance

class Person:
    def __init__(self, name):
        self.name = name
    
    def intro(self):
        print("My name is: ", self.name)

class Employee(Person):
    def __init__(self, name, ID):
        super().__init__(name)
        self.ID = ID
    
    def show_id(self):
        print("Emp ID: ",self.ID)

class Manager(Employee):
    def __init__(self, name, ID, dept):
        super().__init__(name, ID)
        self.dept = dept
    
    def show_dept(self):
        print("Departmet : ", self.dept)

m1 = Manager( "Taqdees Sir", 1, "Devops")
print(m1.name)

m1.intro()
m1.show_id()
m1.show_dept()
```

```python
#Multiple Inheritance
class father:
    def __init__(self, fname):
        self.fname = fname

    def father_name(self):
        print("father name: ", self.fname)

class mother:
    def __init__(self, mname):
        self.mname = mname

    def mother_name(self):
        print("mother name: ", self.mname)

class child(father,mother):
    def __init__(self, fname, mname, cname):
        father.__init__(self,fname)
        mother.__init__(self, mname)
        self.cname = cname
    
    def child_name(self):
        print("child name:",self.cname)


c1 = child("Ashok", "Kajal", "Anirudha")
c1.father_name()
c1.mother_name()
c1.child_name()

print(c1.fname)
```

```python
#Hierrarchical Inheritance
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
    
    def show_brand(self):
        print("Brand:", self.brand)

class Bike(Vehicle):
    def __init__(self, brand, price):
        super().__init__(brand)
        self.price = price
    
    def bike_info(self):
        print("Bike Price: ",self.price)

class car(Vehicle):
    def __init__(self, brand,price):
        super().__init__(brand)
        self.price = price
    
    def car_info(self):
        print("Car price: ", self.price)


b1 = Bike("Honda", 90000)
c1 = car("Toyota", 1200000)

b1.show_brand()
b1.bike_info()

c1.show_brand()
c1.car_info()

```

```python
#Hybrid inheritance

class college:
    def __init__(self, cname):
        self.cname = cname
    
    def college_name(self):
        print("college : ", self.cname)

class student(college):
    def __init__(self,cname,sname):
        super().__init__(cname)
        self.sname = sname
    
    def student_name(self):
        print("student : ",self.sname)

class sports:
    def __init__(self,sport):
        self.sport = sport
    
    def sport_name(self):
        print("Sports : ", self.sport)

class Player(student,sports):
    def __init__(self, cname, sname, sport):
        student.__init__(self, cname, sname)
        sports.__init__(self, sport)
    
    def show_details(self):
        self.college_name()
        self.student_name()
        self.sport_name()

p1 = Player("Bharat College of Engineering", "Anirudha", "cricket")
p1.show_details()
```

```python
#BankManagementSystem

class SBI:
    def __init__(self, accno, balance):
        self.accno = accno
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)
    
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance -= amount
            print("Withdrawn:", amount)
    
    def display_balance(self):
        print("Account No:", self.accno)
        print("Balance:", self.balance)
```

```python
def check_even_list(num_list):
#return all the even numbers in a list
    even_numbers= []
    for number in num_list: #number iterates through num_list we give i/p
        if number%2==0:
            even_numbers.append(number)
        else:
            pass
    return even_numbers
```

```python
def shipping_label(*args, **kwargs):
    for arg in args:
        print(arg, end = " ")
    print()

    print(f"{kwargs.get('apt')}, {kwargs.get('city')}")
    print(f"{kwargs.get('state')}, {kwargs.get('zip')}")
shipping_label("Er.", "Anirudha", "Kolay",
               apt = 108,
               city = "Mumbai",
               state = "maharashtra",
               zip = 421503)
```
